### Data cleaning and preparing for Unsupervised Learning

```python
import pandas as pd

data_path = "data/force/CSV_train.csv"
df = pd.read_csv(data_path, sep=";")
df = df.rename(columns={"FORCE_2020_LITHOFACIES_LITHOLOGY": "LITHOLOGY", "FORCE_2020_LITHOFACIES_CONFIDENCE": "CONFIDENCE"})
df.head()
```

```python
import missingno as msno
import matplotlib.pyplot as plt

#Bar Chart
msno.bar(df)

# Add title
plt.title("Missing Values Bar Chart with Raw data", fontsize=18)
plt.show()
```

Drop columns which have less values or unrelevant columns, including: 'X_LOC', 'Y_LOC', 'Z_LOC','SGR', 'DTS', 'DCAL', 'MUDWEIGHT', 'RMIC', 'ROPA', 'RXO', 'CONFIDENCE'

```python
drop_columns = ['X_LOC', 'Y_LOC', 'Z_LOC','SGR', 'DTS', 'DCAL', 'MUDWEIGHT', 'RMIC', 'ROPA', 'RXO', 'CONFIDENCE']
filtered_df = df.drop(drop_columns, axis=1)
# drop all null values
filtered_df = filtered_df.dropna()

#Bar Chart
msno.bar(filtered_df)

# Add title
plt.title("Missing Values Bar Chart with Raw data", fontsize=18)
plt.show()
```

```python
filtered_df.to_csv("data/force/f_train.csv", index=False)
```
